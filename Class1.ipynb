## ***Task 1 By Teacher***

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from transformers import pipeline

print("--- TASK 1: Preparing Data ---")

training_reviews = [
    "The movie was fantastic",   # 1 = Pos
    "I loved this movie",        # 1 = Pos
    "The movie was terrible",    # 0 = Neg
    "I hated this movie"         # 0 = Neg
]
labels = [1, 1, 0, 0]  # 1 = P, 0 = N

converter = CountVectorizer()
numbers_data = converter.fit_transform(training_reviews)

print("Words learned by computer:")
print(converter.get_feature_names_out())

print("\n--- TASK 2: Training & Testing Bag-of-Words Model ---")

bow_model = LogisticRegression()
bow_model.fit(numbers_data, labels)
print("Model finished learning!\n")

sample_review = "This is  movie fantastic'"


sample_in_numbers = converter.transform([sample_review])


prediction = bow_model.predict(sample_in_numbers)
if prediction[0] == 1:
    print(f"Sample Review: '{sample_review}'")
    print("Result       : Positive")
else:
    print(f"Sample Review: '{sample_review}'")
    print("Result       : Negative")


print("\n--- TASK 3: Side-by-Side Comparison ---")

print("Loading Transformer AI, please wait...")
transformer_ai = pipeline("sentiment-analysis")

test_review = "The movie was not terrible"

# Model 1: Bag-of-Words Prediction
test_numbers = converter.transform([test_review])
bow_prediction = bow_model.predict(test_numbers)

if bow_prediction[0] == 1:
    bow_result = "Positive "
else:
    bow_result = "Negative "

# Model 2: Transformer AI Prediction
transformer_output = transformer_ai(test_review)


transformer_label = transformer_output[0]["label"]

if transformer_label == "POSITIVE":
    transformer_result = "Positive "
else:
    transformer_result = "Negative "

print("---------------------------------------------")
print("Sentence Tested:", test_review)
print("---------------------------------------------")
print("Bag-of-Words Prediction:", bow_result)
print("Transformer Prediction :", transformer_result)
print("---------------------------------------------")

```

## ***Task 1 By Me***

```python

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from transformers import pipeline

print("--- TASK 1: Preparing Data ---")

training_reviews = [
    "The food was amazing",
    "The food was good",
    "I really enjoyed the meal",
    "The food was awful",
    "I disliked the meal"
]

labels = [1, 1, 1, 0, 0]  # 1 = Positive, 0 = Negative

converter = CountVectorizer()
numbers_data = converter.fit_transform(training_reviews)

print("\nWords learned by computer:")
print(converter.get_feature_names_out())

print("\nNumerical representation of reviews:")
print(numbers_data.toarray())

print("\n--- TASK 2: Training Model ---")

model = LogisticRegression()
model.fit(numbers_data, labels)

print("Model trained successfully!")

print("\n--- TASK 3: Testing New Review ---")

new_review = ["The meal was good"]

new_data = converter.transform(new_review)

prediction = model.predict(new_data)

if prediction[0] == 1:
    print("Review:", new_review[0])
    print("Sentiment: Positive")
else:
    print("Review:", new_review[0])
    print("Sentiment: Negative")
```
